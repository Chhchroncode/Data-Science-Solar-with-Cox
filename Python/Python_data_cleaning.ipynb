```python
#Create log column of MW because of skewness in the Capacity MW column
#Ensure linear relationship with Cox hazard and prevent high leverage from a few projects
solar['log_Capacity_MW']=np.log(solar['Capacity_MW'])
print(solar['log_Capacity_MW'].describe())
```

```python
#Add MW band as a way to sort skewness in MW
#MW band are set according to statutory threshold
#10MW:Welsh Threshold for bypassing local council, 50MW: Scottish Threshold for Session 36 planning act, English NSIP threshold before 2025
#100MW: English threshold for NSIP after 2025
#350MW: Welsh Threshold for NSIP, ommitted due to limited projects and events, prevent wide CI in model
#Developers commonly submit project as 9.9, 49.9 etc to maximise within threshold
solar["Capacity_band"] = np.select(
    [solar["Capacity_MW"] <= 9.99,solar["Capacity_MW"] <= 49.99,solar["Capacity_MW"] <= 99.99],
    ["5-9.99 MW", "9.99-49.99 MW","49.99-99.99 MW"], default=">99.99 MW")
print(solar["Capacity_band"].value_counts())
```

```python
#Dropping project that appeal without initial decision, questionable validity
appeal_without_decision=solar[solar[['Secretary of State - Refusal','Secretary of State - Granted','Planning Permission  Granted','Planning Permission Refused']].isna().all(axis=1)&(solar['Appeal Granted'].notna()|solar['Appeal Refused'].notna())]
solar=solar.drop(appeal_without_decision.index)
solar.shape
```

```python
#Add first decision date and final decision date column, and two options for outcome of Cox model
#first decision is more related to modeling the direct time for the authority to decide
#final decision date takes developer's decision to whether appeal into account, more about overall business risk and pipeline timing
#choose to incorporate appeal event within the decision time, and not to use any appeal related data as variable, cox model can fail to converge in relation to refusing outcomes
First_Decision_Cols=['Secretary of State - Refusal','Secretary of State - Granted','Planning Permission  Granted','Planning Permission Refused','Planning Application Withdrawn']
End_Cols=['Secretary of State - Refusal','Secretary of State - Granted','Planning Permission  Granted','Planning Permission Refused','Planning Application Withdrawn','Appeal Granted','Appeal Refused','Appeal Withdrawn']
solar["First_Decision_Date"]=solar[First_Decision_Cols].min(axis=1)
solar["End_Date"]=solar[End_Cols].max(axis=1)
print(solar.isna().sum())
```

```python
#Use the upload date of data as the end_date for projects without decision
#Cox model drops null value in Rstudio
Cutoff_date= pd.to_datetime('04/08/2026',format='%d/%m/%Y')
solar["First_Decision_Date"]=solar["First_Decision_Date"].fillna(Cutoff_date)
solar["End_Date"]=solar["End_Date"].fillna(Cutoff_date)
print(solar.isna().sum())
```

```python
#Event columns indicate an application is consented, refused or withdrawn, if no date, its pending
#First Decision
solar['First_Outcome']='Pending'
has_first_dates = solar[First_Decision_Cols].notna().any(axis=1)
solar.loc[has_first_dates,'First_Outcome']=solar.loc[has_first_dates,First_Decision_Cols].idxmax(axis=1)
print(solar['First_Outcome'].value_counts())
```

```python
solar['First_Outcome']=solar['First_Outcome'].replace({'Secretary of State - Granted':'Planning Permission Granted', 
                                                       'Planning Permission  Granted':'Planning Permission Granted',
                                                       'Secretary of State - Refusal':'Planning Permission Refused'})
print(solar['First_Outcome'].value_counts())
```

```python
#Final Decision
solar['Latest_Outcome']='Pending'
has_latest_dates = solar[End_Cols].notna().any(axis=1)
solar.loc[has_latest_dates,'Latest_Outcome']=solar.loc[has_latest_dates,End_Cols].idxmax(axis=1)
print(solar['Latest_Outcome'].value_counts())
```

```python
solar.loc[solar['Development Status']=='Appeal Lodged',['Latest_Outcome']]='Pending'
print(solar['Latest_Outcome'].value_counts())
```

```python
#Treating appeal withdrawn as withdrawn as the final outcome, instead of refused. As the withdrawnal takes in additional factors than regular refusal as developers make the decision 
solar['Latest_Outcome']=solar['Latest_Outcome'].replace({'Secretary of State - Granted':'Planning Permission Granted', 
                                                       'Planning Permission  Granted':'Planning Permission Granted',
                                                       'Secretary of State - Refusal':'Planning Permission Refused',
                                                        'Appeal Refused':'Planning Permission Refused',
                                                        'Appeal Granted':'Planning Permission Granted',
                                                        'Appeal Withdrawn':'Planning Application Withdrawn'})
print(solar['Latest_Outcome'].value_counts())
```

```python
#First Decision Code
Outcome_Code={'Pending':0,'Planning Permission Granted':1,'Planning Permission Refused':2,'Planning Application Withdrawn':3}
solar['First_Outcome_Code']=solar['First_Outcome'].map(Outcome_Code)
print(solar['First_Outcome_Code'].value_counts())
```

```python
#Latest Decision Code
solar['Latest_Outcome_Code']=solar['Latest_Outcome'].map(Outcome_Code)
print(solar['Latest_Outcome_Code'].value_counts())
```
