```python
#Dropping columns with info irrelevant to solar, and reference numbers that will not be used in analysis
#Dropping columns with detailed location information, GIS is not within scope of this project
#Dropping columns with info on post-consent stages, this  project focuses on consent time
solar=solar.drop(columns=["Operator (or Applicant)","Storage Co-location REPD Ref ID","Turbine Capacity (MW)","No. of Turbines","Height of Turbines (m)","Address","Post Code","X-coordinate","Y-coordinate","Planning Application Reference","Appeal Reference","Secretary of State Reference","Offshore Wind Round","Heat Network Ref","Under Construction","Operational"])
```

```python
print(solar["Mounting Type for Solar"].value_counts())
```

```python
#Rooftop solar is subjected to a different consent regime in prior approval and planning timelines
#Scope of this project focuses on ground mount
solar=solar[solar["Mounting Type for Solar"]=="Ground"]
solar.shape
```

```python
print(solar.isnull().sum())
```

```python
solar["Installed Capacity (MWelec)"] = pd.to_numeric(solar["Installed Capacity (MWelec)"], errors="coerce")
```

```python
#Focus on projects >=5MW, definition of large scale ground mount for DESNZ
#Smaller solar farm face faster consent time, aim to build model that generalise to larger projects in this analysis
solar5=solar[solar["Installed Capacity (MWelec)"]>=5]
solar5.shape
```
